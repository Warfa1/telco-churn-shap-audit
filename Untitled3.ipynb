```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score)
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline

# 1. Split features and target
X = df.drop(columns=['Churn'])
y = df['Churn']

# 2. Stratified Train/Test Split (80/20)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

print(f"Train size: {len(X_train)}, Test size: {len(X_test)}")
print(f"Train churn rate: {y_train.mean()*100:.2f}%")
print(f"Test churn rate:  {y_test.mean()*100:.2f}%")

# 3. Preprocessing (One-Hot for categoricals, Scale for numerics)
cat_cols = X.select_dtypes(include='object').columns.tolist()
num_cols = [c for c in X.columns if c not in cat_cols]

preprocess = ColumnTransformer([
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols),
    ('num', StandardScaler(), num_cols)
])
print("Preprocessing pipeline ready.")
```

```python
# Evaluation helper function
def evaluate_model(model, X_tr, y_tr, X_te, y_te, name):
    model.fit(X_tr, y_tr)
    pred = model.predict(X_te)
    proba = model.predict_proba(X_te)[:, 1]
    return {
        'Model': name,
        'Accuracy': round(accuracy_score(y_te, pred), 3),
        'Precision': round(precision_score(y_te, pred, zero_division=0), 3),
        'Recall': round(recall_score(y_te, pred), 3),
        'F1-Score': round(f1_score(y_te, pred), 3),
        'ROC-AUC': round(roc_auc_score(y_te, proba), 3)
    }

results = []

# --- Baseline 1: Majority Class (always predict "No Churn") ---
results.append(evaluate_model(
    DummyClassifier(strategy='most_frequent'),
    X_train, y_train, X_test, y_test, 'Baseline (Majority)'))

# --- Baseline 2: LR and RF with NO imbalance handling ---
lr_none = Pipeline([('prep', preprocess), ('clf', LogisticRegression(max_iter=2000, random_state=42))])
rf_none = Pipeline([('prep', preprocess), ('clf', RandomForestClassifier(n_estimators=300, random_state=42))])

results.append(evaluate_model(lr_none, X_train, y_train, X_test, y_test, 'LogReg (No Imbalance)'))
results.append(evaluate_model(rf_none, X_train, y_train, X_test, y_test, 'RandomForest (No Imbalance)'))
```

```python
# --- Strategy 1: class_weight='balanced' ---
lr_cw = Pipeline([('prep', preprocess),
                  ('clf', LogisticRegression(max_iter=2000, class_weight='balanced', random_state=42))])
rf_cw = Pipeline([('prep', preprocess),
                  ('clf', RandomForestClassifier(n_estimators=300, class_weight='balanced', random_state=42))])

results.append(evaluate_model(lr_cw, X_train, y_train, X_test, y_test, 'LogReg (Class Weight)'))
results.append(evaluate_model(rf_cw, X_train, y_train, X_test, y_test, 'RandomForest (Class Weight)'))

# --- Strategy 2: SMOTE applied ONLY inside training (Correct) ---
lr_smote = ImbPipeline([('prep', preprocess), ('smote', SMOTE(random_state=42)),
                        ('clf', LogisticRegression(max_iter=2000, random_state=42))])
rf_smote = ImbPipeline([('prep', preprocess), ('smote', SMOTE(random_state=42)),
                        ('clf', RandomForestClassifier(n_estimators=300, random_state=42))])

results.append(evaluate_model(lr_smote, X_train, y_train, X_test, y_test, 'LogReg (SMOTE Correct)'))
results.append(evaluate_model(rf_smote, X_train, y_train, X_test, y_test, 'RandomForest (SMOTE Correct)'))

print("Models trained. Moving to leaky SMOTE demonstration next.")
```

```python
# --- Strategy 3: LEAKY SMOTE (applied BEFORE splitting) ---
# We emulate this by combining train+test, applying SMOTE, then splitting.
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE as SMOTE_leak

# Combine train + test
X_all = pd.concat([X_train, X_test], ignore_index=True)
y_all = pd.concat([y_train, y_test], ignore_index=True)

# Fit preprocessor on all data and transform
prep_fitted = preprocess.fit(X_all, y_all)
X_all_t = prep_fitted.transform(X_all)

# Apply SMOTE on the ENTIRE dataset (the leakage!)
X_all_s, y_all_s = SMOTE_leak(random_state=42).fit_resample(X_all_t, y_all)

# Split AFTER resampling
Xtr_l, Xte_l, ytr_l, yte_l = train_test_split(
    X_all_s, y_all_s, test_size=0.2, stratify=y_all_s, random_state=42)

# Train and evaluate
lr_leak = LogisticRegression(max_iter=2000, random_state=42).fit(Xtr_l, ytr_l)
rf_leak = RandomForestClassifier(n_estimators=300, random_state=42).fit(Xtr_l, ytr_l)

for m, nm in [(lr_leak, 'LogReg (SMOTE LEAKY)'), (rf_leak, 'RandomForest (SMOTE LEAKY)')]:
    p = m.predict(Xte_l)
    pr = m.predict_proba(Xte_l)[:, 1]
    results.append({
        'Model': nm,
        'Accuracy': round(accuracy_score(yte_l, p), 3),
        'Precision': round(precision_score(yte_l, p, zero_division=0), 3),
        'Recall': round(recall_score(yte_l, p), 3),
        'F1-Score': round(f1_score(yte_l, p), 3),
        'ROC-AUC': round(roc_auc_score(yte_l, pr), 3)
    })

# --- Final Table ---
res_df = pd.DataFrame(results)
print("\n=== FULL RESULTS TABLE ===")
print(res_df.to_string(index=False))

# Save to Drive
res_df.to_csv(f'{project_path}/results/metrics_table.csv', index=False)
print(f"\nSaved to: {project_path}/results/metrics_table.csv")
```

import itertools
from scipy.stats import spearmanr

def check_stability(model_class, model_name):
    rankings = []
    print(f"Calculating stability for {model_name} across 5 seeds...")
    
    for seed in [1, 2, 3, 4, 5]:
        # Rebuild and fit the model with a new seed
        m = Pipeline([('prep', preprocess), ('clf', model_class(random_state=seed))])
        m.fit(X_train, y_train)
        
        # Transform data
        Xt = m.named_steps['prep'].transform(X_train)
        fn = m.named_steps['prep'].get_feature_names_out()
        
        # Sample 500 rows for speed
        np.random.seed(42)
        idx = np.random.choice(Xt.shape[0], 500, replace=False)
        X_sample = Xt[idx]
        
        # Calculate SHAP
        if model_name == 'LR':
            sv = shap.LinearExplainer(m.named_steps['clf'], X_sample).shap_values(X_sample)
        else:
            sv_raw = shap.TreeExplainer(m.named_steps['clf']).shap_values(X_sample)
            if isinstance(sv_raw, list):
                sv = sv_raw[1]
            elif len(sv_raw.shape) == 3:
                sv = sv_raw[:, :, 1]
            else:
                sv = sv_raw
                
        # Store the absolute SHAP values
        rankings.append(pd.Series(np.abs(sv).mean(axis=0), index=fn))
    
    # Calculate pairwise Spearman correlation across the 5 seeds

```python
import itertools
from scipy.stats import spearmanr

def check_stability_fast(model_class, model_name):
    rankings = []
    print(f"Calculating stability for {model_name} (3 seeds)...")

    for seed in [1, 2, 3]:
        m = Pipeline([('prep', preprocess), ('clf', model_class(random_state=seed))])
        m.fit(X_train, y_train)
        Xt = m.named_steps['prep'].transform(X_train)
        fn = m.named_steps['prep'].get_feature_names_out()

        np.random.seed(42)
        idx = np.random.choice(Xt.shape[0], 500, replace=False)
        X_sample = Xt[idx]

        if model_name == 'LR':
            sv = shap.LinearExplainer(m.named_steps['clf'], X_sample).shap_values(X_sample)
        else:
            sv_raw = shap.TreeExplainer(m.named_steps['clf']).shap_values(X_sample)
            if isinstance(sv_raw, list):
                sv = sv_raw[1]
            elif len(sv_raw.shape) == 3:
                sv = sv_raw[:, :, 1]
            else:
                sv = sv_raw

        rankings.append(pd.Series(np.abs(sv).mean(axis=0), index=fn))

    corrs = []
    for i, j in itertools.combinations(range(len(rankings)), 2):
        c, _ = spearmanr(rankings[i], rankings[j])
        corrs.append(c)
    return np.mean(corrs)

lr_stability = check_stability_fast(lambda **kw: LogisticRegression(max_iter=2000, class_weight='balanced'), 'LR')
rf_stability = check_stability_fast(lambda **kw: RandomForestClassifier(n_estimators=100, class_weight='balanced'), 'RF')

print(f"\nLR Stability Score: {lr_stability:.3f}")
print(f"RF Stability Score: {rf_stability:.3f}")
```
